````python

import pandas as pd
df_clean = df.copy()
df_clean.columns = df_clean.columns.str.strip()
duplicates_before = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()
for col in ["Order Date", "Ship Date"]:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(
            df_clean[col], errors="coerce"
        )
numeric_columns = ["Sales", "Profit", "Quantity", "Discount"]

for col in numeric_columns:
    if col in df_clean.columns:
        df_clean[col] = pd.to_numeric(
            df_clean[col], errors="coerce"
        )
print("CLEANING SUMMARY")
print("----------------")
print("Original rows:", len(df))
print("Duplicate rows found:", duplicates_before)
print("Rows after duplicate removal:", len(df_clean))
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
display(df_clean.head())

````

````python
total_revenue = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_quantity = df_clean["Quantity"].sum()

profit_margin = (
    total_profit / total_revenue * 100
    if total_revenue != 0 else 0
)
print("BUSINESS PERFORMANCE SUMMARY")
print("=" * 35)
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Unique Orders: {total_orders:,}")
print(f"Units Sold: {total_quantity:,}")
print(f"Profit Margin: {profit_margin:.2f}%")
````

````python

region_performance = (
    df_clean.groupby("Region")
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
    .sort_values("Revenue", ascending=False)
)

region_performance["Profit Margin (%)"] = (
    region_performance["Profit"]
    / region_performance["Revenue"] * 100
)

display(region_performance.round(2))
````

````python

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display, HTML

# Use the cleaned dataset you already created
dashboard_df = df_clean.copy()

# Prepare dates and numeric columns
dashboard_df["Order Date"] = pd.to_datetime(
    dashboard_df["Order Date"], errors="coerce"
)

for col in ["Sales", "Profit", "Quantity", "Discount"]:
    if col in dashboard_df.columns:
        dashboard_df[col] = pd.to_numeric(
            dashboard_df[col], errors="coerce"
        )

dashboard_df = dashboard_df.dropna(
    subset=["Order Date", "Sales", "Profit"]
)

# Calculate business KPIs
revenue = dashboard_df["Sales"].sum()
profit = dashboard_df["Profit"].sum()
orders = dashboard_df["Order ID"].nunique()
units = dashboard_df["Quantity"].sum()
margin = profit / revenue * 100 if revenue else 0

# Display KPI cards
cards = f"""
<div style="display:grid;grid-template-columns:repeat(4,minmax(130px,1fr));gap:12px;">
  <div style="padding:18px;border:1px solid #ddd;border-radius:12px;">
    <div style="color:#666;">Total Revenue</div>
    <h2>${revenue:,.0f}</h2>
  </div>
  <div style="padding:18px;border:1px solid #ddd;border-radius:12px;">
    <div style="color:#666;">Total Profit</div>
    <h2>${profit:,.0f}</h2>
  </div>
  <div style="padding:18px;border:1px solid #ddd;border-radius:12px;">
    <div style="color:#666;">Total Orders</div>
    <h2>{orders:,}</h2>
  </div>
  <div style="padding:18px;border:1px solid #ddd;border-radius:12px;">
    <div style="color:#666;">Profit Margin</div>
    <h2>{margin:.2f}%</h2>
  </div>
</div>
"""

display(HTML("<h1>Business Operations & KPI Dashboard</h1>"))
display(HTML(cards))

print(f"Units sold: {units:,.0f}")

# 1. Monthly revenue and profit
monthly = (
    dashboard_df.assign(
        Month=dashboard_df["Order Date"].dt.to_period("M").astype(str)
    )
    .groupby("Month", as_index=False)
    .agg(Revenue=("Sales", "sum"), Profit=("Profit", "sum"))
)

fig_monthly = px.line(
    monthly,
    x="Month",
    y=["Revenue", "Profit"],
    markers=True,
    title="Monthly Revenue and Profit Trends",
    labels={"value": "Amount ($)", "variable": "Metric"}
)
fig_monthly.update_layout(hovermode="x unified")
fig_monthly.show()

# 2. Regional performance
region = (
    dashboard_df.groupby("Region", as_index=False)
    .agg(
        Revenue=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
)

fig_region = px.bar(
    region.sort_values("Revenue", ascending=False),
    x="Region",
    y="Revenue",
    color="Profit",
    title="Revenue by Region",
    hover_data=["Profit", "Orders"],
    color_continuous_scale="Blues"
)
fig_region.show()

# 3. Product category performance
category = (
    dashboard_df.groupby("Category", as_index=False)
    .agg(Revenue=("Sales", "sum"), Profit=("Profit", "sum"))
)

fig_category = px.bar(
    category,
    x="Category",
    y=["Revenue", "Profit"],
    barmode="group",
    title="Revenue vs Profit by Product Category",
    labels={"value": "Amount ($)", "variable": "Metric"}
)
fig_category.show()

# 4. Most and least profitable subcategories
subcategory = (
    dashboard_df.groupby("Sub-Category", as_index=False)
    .agg(Profit=("Profit", "sum"), Revenue=("Sales", "sum"))
    .sort_values("Profit")
)

fig_products = px.bar(
    subcategory,
    x="Profit",
    y="Sub-Category",
    orientation="h",
    color="Profit",
    title="Profitability by Product Subcategory",
    color_continuous_scale="RdYlGn",
    hover_data=["Revenue"]
)
fig_products.show()

print("Dashboard overview generated successfully!")
````

````python

import os
import zipfile

project = "business-operations-kpi-dashboard"
os.makedirs(project, exist_ok=True)

files_to_create = {
    "analysis.py": r'''
import pandas as pd

def load_data(path="superstore_cleaned.csv"):
    df = pd.read_csv(path, encoding="utf-8-sig")
    df.columns = df.columns.str.strip()

    df["Order Date"] = pd.to_datetime(
        df["Order Date"], errors="coerce"
    )

    for col in ["Sales", "Profit", "Quantity", "Discount"]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    return df

def calculate_kpis(df):
    revenue = df["Sales"].sum()
    profit = df["Profit"].sum()

    return {
        "Revenue": revenue,
        "Profit": profit,
        "Orders": df["Order ID"].nunique(),
        "Profit Margin (%)": (
            profit / revenue * 100 if revenue else 0
        )
    }

if __name__ == "__main__":
    df = load_data()
    print("Dataset shape:", df.shape)
    print("Business KPIs:")
    for key, value in calculate_kpis(df).items():
        print(f"{key}: {value:,.2f}")
''',

    "app.py": r'''
import streamlit as st
import plotly.express as px
from analysis import load_data, calculate_kpis

st.set_page_config(
    page_title="Business Operations KPI Dashboard",
    layout="wide"
)

st.title("Business Operations & KPI Dashboard")
st.caption("Retail sales performance and profitability analysis")

try:
    df = load_data()
except FileNotFoundError:
    st.error(
        "Dataset not found. Place superstore_cleaned.csv "
        "in the project directory. See README.md."
    )
    st.stop()

regions = ["All"] + sorted(df["Region"].dropna().unique().tolist())
categories = ["All"] + sorted(
    df["Category"].dropna().unique().tolist()
)

c1, c2 = st.columns(2)
with c1:
    selected_region = st.selectbox("Region", regions)
with c2:
    selected_category = st.selectbox("Category", categories)

filtered = df.copy()

if selected_region != "All":
    filtered = filtered[filtered["Region"] == selected_region]

if selected_category != "All":
    filtered = filtered[filtered["Category"] == selected_category]

if filtered.empty:
    st.warning("No data matches the selected filters.")
    st.stop()

kpis = calculate_kpis(filtered)

a, b, c, d = st.columns(4)
a.metric("Total Revenue", f"${kpis['Revenue']:,.2f}")
b.metric("Total Profit", f"${kpis['Profit']:,.2f}")
c.metric("Distinct Orders", f"{kpis['Orders']:,}")
d.metric("Profit Margin", f"{kpis['Profit Margin (%)']:.2f}%")

monthly = filtered.dropna(subset=["Order Date"]).copy()
monthly["Month"] = monthly["Order Date"].dt.to_period("M").astype(str)
monthly = monthly.groupby("Month", as_index=False).agg(
    Revenue=("Sales", "sum"),
    Profit=("Profit", "sum")
)

st.subheader("Monthly Performance")
st.plotly_chart(
    px.line(
        monthly, x="Month", y=["Revenue", "Profit"],
        markers=True, title="Revenue and Profit by Month"
    ),
    use_container_width=True
)

left, right = st.columns(2)

region_data = filtered.groupby("Region", as_index=False).agg(
    Revenue=("Sales", "sum"),
    Profit=("Profit", "sum")
)
with left:
    st.plotly_chart(
        px.bar(
            region_data, x="Region", y="Revenue",
            title="Revenue by Region"
        ),
        use_container_width=True
    )

product_data = filtered.groupby("Sub-Category", as_index=False).agg(
    Revenue=("Sales", "sum"),
    Profit=("Profit", "sum")
).sort_values("Profit")

with right:
    st.plotly_chart(
        px.bar(
            product_data, x="Profit", y="Sub-Category",
            orientation="h", title="Profit by Subcategory"
        ),
        use_container_width=True
    )

st.subheader("Category Performance")
category_data = filtered.groupby("Category", as_index=False).agg(
    Revenue=("Sales", "sum"),
    Profit=("Profit", "sum")
)
st.plotly_chart(
    px.bar(
        category_data, x="Category", y=["Revenue", "Profit"],
        barmode="group", title="Revenue vs. Profit"
    ),
    use_container_width=True
)

st.subheader("Regional KPI Summary")
summary = filtered.groupby("Region", as_index=False).agg(
    Revenue=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Orders=("Order ID", "nunique")
)
summary["Profit Margin (%)"] = (
    summary["Profit"] / summary["Revenue"] * 100
)
st.dataframe(summary.round(2), use_container_width=True)
''',

    "business_kpi_queries.sql": r'''
-- Run these queries against a SQLite table named sales.

-- Overall business KPIs
SELECT
    ROUND(SUM(Sales), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit,
    COUNT(DISTINCT "Order ID") AS Orders,
    ROUND(100.0 * SUM(Profit) / NULLIF(SUM(Sales), 0), 2)
        AS Profit_Margin_Percent
FROM sales;

-- Regional performance
SELECT
    Region,
    ROUND(SUM(Sales), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit
FROM sales
GROUP BY Region
ORDER BY Revenue DESC;

-- Product profitability
SELECT
    "Sub-Category",
    ROUND(SUM(Sales), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit
FROM sales
GROUP BY "Sub-Category"
ORDER BY Profit ASC;

-- Monthly performance
SELECT
    strftime('%Y-%m', "Order Date") AS Month,
    ROUND(SUM(Sales), 2) AS Revenue,
    ROUND(SUM(Profit), 2) AS Profit
FROM sales
WHERE "Order Date" IS NOT NULL
GROUP BY strftime('%Y-%m', "Order Date")
ORDER BY Month;
''',

    "requirements.txt": r'''
pandas>=2.0
streamlit>=1.30
plotly>=5.18
''',

    ".gitignore": r'''
__pycache__/
.venv/
venv/
.env
*.pyc
.DS_Store
superstore_cleaned.csv
*.ipynb
''',

    "business_insights.md": r'''
# Business Insights

## Validated Dataset Results
- Records: 9,994
- Columns: 21
- Total revenue: $2,297,200.86
- Total profit: $286,397.02
- Distinct orders: 5,009

## Regional Findings
- West has the highest revenue and profit in the initial analysis.
- Central has the lowest regional profit margin at 7.92%.

## Recommendations to Investigate
1. Investigate product mix, discounting and cost patterns in
   the Central region.
2. Compare product profitability across regions.
3. Evaluate discounts alongside profit, not just revenue.
4. Monitor revenue and profit margin together.

These are exploratory findings, not proof of causation.
''',

    "README.md": r'''
# Business Operations & KPI Dashboard

A retail business analytics project using Python, SQL and an
interactive Streamlit dashboard.

## Features
- Revenue, profit, order and profit-margin KPIs
- Monthly revenue and profit trends
- Regional performance analysis
- Product-subcategory profitability
- Region and category filters
- SQL queries for business analysis

## Technology Stack
Python, Pandas, SQLite SQL, Plotly and Streamlit.

## Dataset
This project uses the Sample Superstore dataset. Obtain the
dataset from an authorized source and follow its license.

For local execution, save your cleaned CSV as
`superstore_cleaned.csv` in the project directory.

## Run Locally
Install Python, then run:

```bash
pip install -r requirements.txt
streamlit run app.py
```

Open the local URL displayed by Streamlit in your browser.

## SQL
The file `business_kpi_queries.sql` contains example queries.
Load the CSV into a SQLite table named `sales` before running
the queries.

## Results
- 9,994 records in the validated dataset
- Revenue: $2,297,200.86
- Profit: $286,397.02
- 5,009 distinct orders

## Limitations
The dataset is historical sample retail data. Findings describe
this dataset and should not be interpreted as current company
performance. Associations do not prove causation.

## License
Add an appropriate license after confirming rights to the code
and dataset. Do not redistribute data without permission.
'''
}

for filename, content in files_to_create.items():
    with open(os.path.join(project, filename), "w", encoding="utf-8") as f:
        f.write(content.strip() + "\n")

zip_name = project + ".zip"

with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as z:
    for filename in files_to_create:
        path = os.path.join(project, filename)
        z.write(path, arcname=f"{project}/{filename}")

print("Created project files:")
for filename in files_to_create:
    print("-", filename)

from google.colab import files
files.download(zip_name)
````

````python

import os
import zipfile
from google.colab import files

project = "business-operations-kpi-dashboard"
zip_name = project + ".zip"

if not os.path.isdir(project):
    raise FileNotFoundError(
        f"Folder '{project}' not found. "
        "Run the cell that created your project files first."
    )

with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as z:
    for filename in os.listdir(project):
        filepath = os.path.join(project, filename)

        if os.path.isfile(filepath):
            z.write(
                filepath,
                arcname=f"{project}/{filename}"
            )

print("ZIP created:", zip_name)
print("Files included:")

with zipfile.ZipFile(zip_name, "r") as z:
    for filename in z.namelist():
        print("-", filename)

files.download(zip_name)
````
